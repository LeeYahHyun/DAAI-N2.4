```python
placeholders = ['N/A', 'NA', 'n/a', 'null', 'None', 'none', '-', '--', '?',
                 'unknown', 'Unknown', 'nan', '#N/A', 'missing']

mask_placeholder = df['review_title'].astype(str).str.strip().isin(placeholders)
print("Số dòng dùng placeholder giả:", mask_placeholder.sum())
print(df.loc[mask_placeholder, 'review_title'].value_counts())
```

## 04 -  Duplicates & Invalid Data

```python
df[df.duplicated()]
```

```python
df.duplicated().sum()
```

```python
# Loại bỏ cột khóa chính ra khỏi việc so sánh
cols_to_check = [c for c in df.columns if c != 'return_id']

duplicates = df[df.duplicated(subset=cols_to_check, keep=False)]
print("Số dòng trùng nội dung (khác id):", len(duplicates))
duplicates.sort_values(cols_to_check)
```

```python
df[df['rating'] <= 0]
```

```python
# Không âm, không có giá trị lạ
print((df['order_id'] < 0).sum())
print((df['product_id'] < 0).sum())
print((df['customer_id'] < 0).sum())

# Có bị trùng bất thường không (tùy ngữ cảnh — vd: customer_id trùng là bình thường vì 1 khách mua nhiều lần)
print(df['order_id'].duplicated().sum())
```
