##  05 - Outliers

```python
df['discount_value'].describe()
# Khong co dau hieu bat thường
```

```python
df['min_order_value'].describe()
#Bat thuong tuy nhien, day la gia tri toi thieu can ktr them
```

```python
# Xem phân bố cụ thể — bao nhiêu dòng có giá trị 0
print(df['min_order_value'].value_counts().sort_index())

# Xem % dòng = 0
print((df['min_order_value'] == 0).mean() * 100, "%")
```

```python
print("Giá trị âm:", (df['min_order_value'] < 0).sum())

Q1 = df['min_order_value'].quantile(0.25)
Q3 = df['min_order_value'].quantile(0.75)
IQR = Q3 - Q1
upper = Q3 + 1.5 * IQR
print("Ngưỡng trên (IQR):", upper)
print("Số dòng vượt ngưỡng:", (df['min_order_value'] > upper).sum())
```
